## Step 2: Data Cleaning Process

- Removing Irrelevant columns

```python
df.drop(columns =['Unnamed: 0'], inplace= True)
```

 - Missing Value Percentage Analysis

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

- Missing Data Detection

```python
df.isnull().sum()
```

Observed Missing Columns:-

 - Bedrooms
- Bathrooms

###Missing Data Treatment

```python
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].mean())
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].mean())
```

```python
df.isna().sum()
```

###Data Formatting.

```python
#Panda DateTime Conversion
df['date'] = pd.to_datetime(df['date'])
```

```python
df['date'].dtype
```

##Step 3: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

###Q1.What is the distrubution of house prices?

```python
#Distribution of House Prices
plt.figure()
plt.hist(df['price'], bins=30, color='red',alpha=0.8)
plt.title("Distrubution of House Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

####Analysis:
- House prices are mostly concentrated in the lower and middle ranges.
- A few houses have very high prices

####Conclusion:
- Expensive luxury houses are fewer.
- Most properties are in the affordable price range.

###Q2. How does living area affect price?

```python
#Relationship Between Living Area and House Price
plt.scatter(df['sqft_living'], df['price'],color ='red',alpha=0.8)
plt.xlabel("Sqft of Living")
plt.ylabel("Price")
plt.title("Price vc Living Area")
plt.show()
```

####Analysis:
- Living area and house price have a strong positive relationship.
####Conclusion:
- Bigger houses generally have higher prices.

###Q3. What is the relationship between bedrooms and Price.

```python
#Relationship Between Number of Bedrooms and House Price
sns.boxplot(x = 'bedrooms', y='price', data=df,color='red')
plt.title("Bedrooms vc Price")
plt.show()
```

####Analysis:
- House prices increase as the number of bedrooms rises, up to a certain point.
####Conclusion:
- After 6 bedrooms, house price growth appears to slow down.

###Q4. Do bathrooms impack house price?

```python
#Relationship Between Bathrooms and House Price
sns.scatterplot(x ='bathrooms', y ='price', data = df,color='red')
plt.title("Price vc Bathrooms")
plt.show()
```

####Analysis:
- Bathrooms show a stronger effect on house prices than bedrooms.
####Conclusion:
- The number of bathrooms may be a more important factor in house pricing than the number of bedrooms.

###Q5.Does waterfront view increase price?

```python
#Impact of Waterfront on House Prices
sns.boxplot(x='waterfront', y='price', data=df,color='red')
plt.title("Waterfront vs Price")
plt.show()
```

####Analysis:
- Houses with waterfront access generally have higher prices.
####Conclusion:
- Waterfront location is a premium feature that can increase a house's value.

###Q6.What is the correlation among key variables?

```python
#Correlation Analysis Using Heatmap
corr = df[['price','sqft_living','bathrooms','bedrooms','grade']].corr()
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

####Analysis
- sqft_living has strongest correlation
####Conclusion
- Living space is the best price predictor

###Q7. Correlation Heatmap of House Features

```python
# Q7: Correlation Heatmap
corr = df[['price', 'sqft_lot', 'floors',
           'condition', 'yr_built']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Between House Features")
plt.show()
```


####Analysis:
- The heatmap shows the relationships between house prices, lot size, floors, condition, and year built.
- Each correlation value indicates the strength and direction of the relationship between two features.


####Conclusion:

 - The heatmap helps identify relationships between different house features and understand which factors may influence house prices.
